# Parameters

## By Referencde immutable

```python
def cube(x):
    print("Received ", x, "id ", id(x))
    x = x**3
    print("X inside", x, "id", id(x))
    return x

x = 3
print("X before", x, "id x", id(x))
z = cube(x)
print("X after ", x, "id", id(x))
print("z", z, "id", id(z))
```

## By reference mutable

```python
def duplicate(elements: list):
    print("Received ", elements, "id ", id(elements))
    elements.append("a")
    elements.append("b")
    print("Changed ", elements, "id ", id(elements))
    return elements

elements = ["1", "2", "3"]
print("elements before", elements, "id x", id(elements))
l1 = duplicate(elements)
print("elements after ", elements, "id", id(elements))
print("l1", l1, "id", id(l1))
```
