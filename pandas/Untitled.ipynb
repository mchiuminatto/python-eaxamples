```python
import pandas as pd
```

```python
ds_name = "CANDLE_EURUSD_H1_bid_UTC.csv"
ts_name = "EURUSD_Ticks_2010.01.01_2021.04.09.csv"
```

```python
for df in pd.read_csv(ds_name,  chunksize=100):
    df["date_time"] = pd.to_datetime(df["date_time"])
    df["end_time"] = pd.to_datetime(df["end_time"])
    
    _date_from = df.iloc[0]["date_time"]
    _date_to = df.iloc[-1]["end_time"]
    
    df_tk = pd.read_csv(ts_name, index_col="date_time", engine="c", skiprows=lambda x:x not in [_date_from,_date_to])
    print(len(df_tk))
```
