```python
import pandas as pd
import numpy as np
```

```python
# generate a m1 time series

rng = pd.date_range('2019-01-01 00:00', periods=60*24*365, freq='min')  # generate the index
df = pd.DataFrame(np.random.randint(2, size=len(rng)), index=rng, columns=["label"])
```

```python
df["label"]=0
mark_from = '2019-01-01 3:45:00'
mark_to = '2019-01-01 4:15:00'
df["label"][mark_from:mark_to] = 1
df["label"][mark_from:mark_to]
```

```python
df_rsm = df.resample("4H").max()
df_rsm
```
