## DataFrames

### Create from Dictionary

```python
d1 = {"PIB":
    {"europe":1, 
      "latin-america": 3, 
      "mena":6, 
      "asia":8
     }
}

df = pd.DataFrame.from_dict(d1, orient="index").T
df
```

### Add Rows

```python
df.loc[len(df)] = [7]
df.rename(index={df.index[-1]:"oceania"}, inplace=True)
df
```

```python
for row in df.iterrows():
    print(row[0], row[1].PIB)
```

```python
for label, content in df.items():
    print("Label", label) 
    print("Content", content)
```

```python
leverage = lambda x, p: x*p
```

```python
df.apply(leverage, args=(15, ))
```
