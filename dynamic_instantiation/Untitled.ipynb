# Dynamic Instantiation

```python
class JFOREX:
    def open(self):
        print("Dukascopy Format")
        return True
    
class MT4:
    def open(self):
        print("MT4 Format")
        return True
    

class Process:
    
    def __init__(self, format):
        if format == "JFOREX":
            self.src = JFOREX()
        elif format == "MT4":
            self.src = MT4()
            
    def process(self):
        self.src.open()
        


_pcs = Process("MT4")
_pcs.process()
        
```
