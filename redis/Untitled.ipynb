### Expire, counters

```python
r.set("chapter", "chapter 1")
r.expire("chapter", 5)

r.set("counter",1)
for i in range(10):
    print(r.get("chapter").decode())
    time.sleep(1)
    print(f"Time left {r.ttl('chapter')}")
    r.incr("counter")
    print(f"Iteration {r.get('counter').decode()}")
```

### Voting System

```python
def up_vote(id):
    key = "article:"+id+":votes"
    r.incr(key)
    
def down_vote(id):
    key = "article:"+id+":votes"
    r.decr(key)
    
def show_resuls(id):
    _head_key = "article:"+id+":headline"
    _vote_key = "article:"+id+":votes"
    _headline = r.get(_head_key).decode()
    _votes = r.get(_vote_key).decode()
    print("The article " + _headline + " has " + _votes + " votes")
    
```

```python
up_vote("60056")
show_resuls("60056")
```

```python
down_vote("60056")
show_resuls("60056")
```
