```python
from itertools import permutations
import pickle
```

```python
class GetId:
    
    def __init__(self):
        self.ids = permutations("abcde", 2)
        
    def get_id(self):
        return ''.join(next(self.ids))
```

```python
idgen = GetId()
```

```python
idgen.get_id()
```

```python
idgen.get_id()
```

```python
idgen.get_id()
```

```python
with open("id_iterator.pkl", "bw") as fp:
    pickle.dump(idgen, fp)
```

```python
with open("id_iterator.pkl", "br") as fp:
    idg_2 = pickle.load(fp)
```
