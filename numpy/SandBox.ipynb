```python
A = np.random.randint(-10, 10, 20)
B = np.random.randint(50, 100, 20)
```

```python
C = np.concatenate(([A], [B]), axis=1)
print(C.shape)
C
```

```python
C = np.concatenate((A, B), axis=0)
print(C.shape)
C
```

```python
C = np.concatenate((A.T, B.T), axis=0)
print(C.shape)
C
```
