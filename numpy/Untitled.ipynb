```python
import numpy as np
```

```python
points = np.arange(-1,1.5, 0.2)
points
```

```python
xs, ys = np.meshgrid(points, points)
```

```python
xs
```

```python
ys
```

```python
z = np.sqrt(xs ** 2 + ys ** 2)
```

```python
import matplotlib.pyplot as plt
```

```python
plt.imshow(z, cmap=plt.cm.gray )
plt.colorbar()
```
