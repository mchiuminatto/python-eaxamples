```python
# no copy

l1= [1, 2, 3]
l2 = ["a", "b", "c"]
l2 = ["a1", "b2", "c3"]

l = [l1, l2, l3]
print(l)
print(id(l1), id(l2), id(l3), id(l))
l1.append(4)

print(l1)
print(l)
print(id(l1), id(l2), id(l3), id(l))

```

```python
# shallow copy
l1= [1, 2, 3]
l2 = ["a", "b", "c"]
l2 = ["a1", "b2", "c3"]

l = [l1, l2, l3]
print("l", l)
print(id(l1), id(l2), id(l3), id(l))
lcopy = l.copy()
print("lcopy", lcopy)
print(id(l1), id(l2), id(l3), id(l), id(lcopy))

l1.append(4)
print("l1", l1)
print("l", l)
print("lcopy", lcopy)
lcopy = l1.copy()
print(id(l1), id(l2), id(l3), id(l), id(lcopy))


```

```python
# deep copy

import copy

l1= [1, 2, 3]
l2 = ["a", "b", "c"]
l2 = ["a1", "b2", "c3"]

l = [l1, l2, l3]
print("l", l)
print(id(l1), id(l2), id(l3), id(l))
lcopy = copy.deepcopy(l)
print("lcopy", lcopy)
print(id(l1), id(l2), id(l3), id(l), id(lcopy))

l1.append(4)
print("l1", l1)
print("l", l)
print("lcopy", lcopy)
lcopy = l1.copy()
print(id(l1), id(l2), id(l3), id(l), id(lcopy))
```
