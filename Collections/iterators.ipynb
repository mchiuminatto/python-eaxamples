## Sort

```python
profits_raw = [1.10, 2.4, 0.5, 2.3, 8.1, 0.3]
sorted(profits_raw)
```

```python
profits_strategy = [("rsi", 0.8), ("cci", 2.3), ("bbands", 4.5), ("macd", -4.5)]
sorted(profits_strategy, key=lambda x: x[1], reverse=True)
```

## Filter

```python
list(filter(lambda x: "i" in x[0], profits_strategy ))
```

## Map

```python
orders = ['cold brew', 'lemongrass tea', 'chai latte', 'medium drip','french press', 
          'mocha cappuccino', 'pumpkin spicelatte','double-shot espresso', 'dark roast drip', 'americano']

def brew(order):
    return " ".join((order, " ready"))

    
for status in map(brew, orders):
    print(status)
```
