```python
import requests
import json
import hashlib

ENDPOINT = "https://api.close.com/buildwithus/"
ENCODING = "utf-8"

def hash_input_array(input_array, enc_key, dig_size=64):
    hasher = hashlib.blake2b(digest_size=dig_size, key=enc_key)
    hash_list = []
    for item in input_array:
        hasher.update(item.encode(ENCODING))
        result: str = hasher.hexdigest()
        hash_list.append(result)
        
    return hash_list


response = requests.get(url=ENDPOINT)
print("header", response.headers)

requirement = json.loads(response.content)

input_array = requirement["traits"]
KEY = requirement["key"].encode(ENCODING)

hash_input_array_result = hash_input_array(requirement["traits"], enc_key=KEY, dig_size=64)  
raw = hash_input_array_result

response = requests.post(url=ENDPOINT, json=raw, headers={"Content-Type": "application/json"})
print(response.reason, response.status_code, response.text, response.headers)


```
