# Functions

## Scopes

### Basic example of scoping

```python
test = 3
print("global test ", test, id)

def f():
    test = 5
    print("local test ", test)

f()
print("global test ", test)
```

### nonlocal clause

```python
# scoping.level.2.nonlocal.py
def outer():
    test = 1  # outer scope
    def inner():
        nonlocal test
        test = 2  # nearest enclosing scope (which is 'outer')
        print('inner:', test)
    inner()
    print('outer:', test)
test = 0  # global scope
outer()
print('global:', test)
```

### global clause

```python
# scoping.level.2.nonlocal.py
def outer():
    test = 1  # outer scope
    def inner():
        global test
        test = 2  
        print('inner:', test)
    inner()
    print('outer:', test)
test = 0  # global scope
outer()
print('global:', test)

```

### Combining parameter scope with immutability

```python
x = 5


def f(y):
    print("inside before assignment", y, id(y))
    y = 7
    print("inside after assignment", y, id(y))

    
print("outside before calling ", x, id(x))    
f(x)
print("outside after calling ", x, id(x))
```

## Parameters

### Passingg Parameters

```python
test = 1
print("test before ", test, id(test))

def f(x):
    x = 5
    

f(test)
print("test after ", test, id(test))
```

```python
test = [1,2,3]
print("test before ", test, id(test))
def f(x):
    x[1] = 100
f(test)
print("test before ", test, id(test))
```

#### Positional Arguments

```python
# arguments.positional.py
def func(a, b, c):
    print(a, b, c)
func(1, 2, 3)  # prints: 1 2 3
```

#### Keyword Arguments

```python
def func(a, b, c):
    print(a, b, c)
func(a=1, c=2, b=3)  # prints: 1 2 3
```

#### Iterable unpacking

```python
# arguments.unpack.iterable.py
def func(a, b, c):
    print(a, b, c)

values = (1, 3, -7)
func(*values)  # equivalent to: func(1, 3, -7)
func(1, 3, -7)
```

### Dictionary Unpacking

```python
# arguments.unpack.dict.py
def func(a, b, c):
    print(a, b, c)
values = {'b': 1, 'c': 2, 'a': 42}
func(**values)  # equivalent to func(b=1, c=2, a=42)
```

```python
# arguments.combined.py
def func(a, b, c, d, e, f):
    print(a, b, c, d, e, f)
    
func(1, *(2, 3), f=6, *(4, 5))
func(*(1, 2), e=5, *(3, 4), f=6)
func(1, **{'b': 2, 'c': 3}, d=4, **{'e': 5, 'f': 6})
func(c=3, *(1, 2), **{'d': 4}, e=5, **{'f': 6})
```

### Types of parameter definition


-   Positional or keyword parameters: allow both positional and keyword arguments
-   Variable positional parameters: collect an arbitrary number of positional arguments in a tuple
-   Variable keyword parameters: collect an arbitrary number of keyword arguments in a dictionary
-   Positional-only parameters: can only be passed as positional arguments
-   Keyword-only parameters: can only be passed as keyword arguments

### Optional Parameters

```python
def calculate_profits(trades: list[(float, float)], pip_factor: int=4) -> float:
    profit: float = 0
    for trade in trades:
        profit += (trade[1] - trade[0]) * 10**pip_factor
    return profit
```

```python
trades = [(1.1300, 1.1305), (1.1310, 1.1295)]
calculate_profits(trades)
```

```python
trades = [(1.1300, 1.1305), (1.1310, 1.1295)]
calculate_profits(trades, pip_factor=2)
```

Optional parameters must be at the right side of positional parameters

```python
def calculate_profits(pip_factor: int=4, trades: list[(float, float)]) -> float:
    profit: float = 0
    for trade in trades:
        profit += (trade[1] - trade[0]) * 10**pip_factor
    return profit
```

### Variable Positional Arguments

```python
def minimum(*n):
    if n:
        _minimum = n[0]
        for number in n[1:]:
            if number < _minimum:
                _minimum = number
    
    return _minimum
            
```

```python
%timeit minimum(1,2,3,4,5,6)
```

### Variable Keywords Arguments

```python
def plot_chart(**kwargs):
    
    if kwargs:
        print("parameters passed")
        for k,v in kwargs.items():
            print(k, v)
```

```python
plot_chart(a=1, b=2, f=3, c=4)
```

#### Positional Only Parameters

```python
def f(a, b, /, c):
    print(a, b, c)
```

```python
f(1,2,3)
f(1,2, c=10)
```

```python
f(1, b=2, c=3)
```

### Rules Summary


* Positional-only parameters come first, followed by a /.
* Normal parameters go after any positional-only parameters.
* Variable positional parameters go after normal parameters.
* Keyword-only parameters go after variable positional parameters.
* Variable keyword parameters always go last.
* For positional-only and normal parameters, any required parameters have to be defined before any optional parameters. This means that if you have an optional positional-only parameter, all your normal parameters must be optional too. This rule doesn't affect keyword-only parameters.

```python
# full example

def f(a, b, c=7, *args, **kwargs):
    print("a, b, c", a, b, c)
    print("args", args)
    print("kwargs", kwargs)
```

```python
f(1,2, 3, 5, 7 , 9, A='a', B='b')
```
